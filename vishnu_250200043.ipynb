```python
def analyze_movies(df):
    """
    Complete movie analysis function
    """
    results = {}

    # 1. Basic statistics
    results['total_movies'] = len(df)
    results['avg_rating'] = df['Rating'].mean()
    results['total_gross'] = df['Gross_Millions'].sum()

    # 2. Top performers
    results['top_movie'] = (
        df.nlargest(1, 'Rating')[['Title', 'Rating']]
        .to_dict(orient='records')
    )

    results['top_director'] = (
        df.groupby('Director')['Rating']
        .mean()
        .idxmax()
    )

    # 3. Yearly analysis
    results['yearly_stats'] = (
        df.groupby('Year')
        .agg({
            'Rating': ['mean', 'count'],
            'Gross_Millions': 'sum'
        })
        .round(2)
    )

    # 4. Genre analysis
    genre_df = df[['Genre', 'Rating']].copy()
    genre_df['Genre'] = genre_df['Genre'].str.split(', ')
    genre_df = genre_df.explode('Genre')

    results['top_genre'] = (
        genre_df['Genre']
        .value_counts()
        .idxmax()
    )

    results['genre_ratings'] = (
        genre_df.groupby('Genre')['Rating']
        .mean()
        .sort_values(ascending=False)
    )

    # 5. Correlations
    results['rating_votes_corr'] = (
        df['Rating'].corr(df['Votes'])
    )

    return results


# Run analysis
analysis_results = analyze_movies(df)


# Print results
print("=== MOVIE DATASET ANALYSIS ===\n")

print(f"Total Movies: {analysis_results['total_movies']}")
print(f"Average Rating: {analysis_results['avg_rating']:.2f}")
print(f"Total Gross: ${analysis_results['total_gross']:.0f} M")
print(f"Top Director: {analysis_results['top_director']}")
print(f"Most Common Genre: {analysis_results['top_genre']}")
print(
    f"Rating-Votes Correlation: "
    f"{analysis_results['rating_votes_corr']:.3f}"
)

print("\nYearly Statistics:")
print(analysis_results['yearly_stats'])

print("\nTop 5 Genres by Rating:")
print(analysis_results['genre_ratings'].head())
```

```python
import pandas as pd


def clean_movie_data(df):
    """
    Comprehensive data cleaning pipeline.
    """

    # Make a copy to avoid modifying the original DataFrame
    df_clean = df.copy()

    # 1. Clean string columns
    df_clean["Title"] = df_clean["Title"].str.strip()
    df_clean["Director"] = df_clean["Director"].str.strip().str.title()
    df_clean["Actor"] = df_clean["Actor"].str.strip()
    df_clean["Genre"] = df_clean["Genre"].str.strip()

    # 2. Clean numeric columns

    # Convert Gross from "$50M" to 50.0
    df_clean["Gross_Millions"] = (
        df_clean["Gross"]
        .str.replace("$", "", regex=False)
        .str.replace("M", "", regex=False)
        .str.strip()
        .astype(float)
    )

    # Convert Runtime from "148 min" to 148
    df_clean["Runtime_Minutes"] = (
        df_clean["Runtime"]
        .str.replace("min", "", regex=False)
        .str.strip()
        .astype(int)
    )

    # 3. Create rating categories
    rating_bins = [0, 7, 8, 9, 10]
    rating_labels = ["Poor", "Average", "Good", "Excellent"]

    df_clean["Rating_Category"] = pd.cut(
        df_clean["Rating"],
        bins=rating_bins,
        labels=rating_labels,
        include_lowest=True
    )

    # 4. Split genres for analysis
    df_clean["Genre_List"] = df_clean["Genre"].str.split(", ")

    # Get the first genre as the primary genre
    df_clean["Primary_Genre"] = df_clean["Genre_List"].str[0]

    # 5. Add derived columns
    df_clean["Rating_Rounded"] = df_clean["Rating"].round(1)

    df_clean["Votes_Millions"] = (
        df_clean["Votes"] / 1_000_000
    ).round(2)

    # 6. Handle missing values
    df_clean["Rating_Category"] = (
        df_clean["Rating_Category"]
        .cat.add_categories("Not Rated")
        .fillna("Not Rated")
    )

    df_clean["Gross_Millions"] = (
        df_clean["Gross_Millions"]
        .fillna(df_clean["Gross_Millions"].median())
    )

    return df_clean


# Apply cleaning
df_cleaned = clean_movie_data(df)

# Display information about the cleaned DataFrame
print("Cleaned DataFrame Info:")
df_cleaned.info()

print("\nFirst row after cleaning:")
print(df_cleaned.iloc[0])
```
